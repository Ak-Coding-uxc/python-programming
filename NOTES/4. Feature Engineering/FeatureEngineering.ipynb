## LabelEncoder

Ye class text ko numbers me convert karti hai. Used in Ordered Data.

```python
le = LabelEncoder()
```

```python
company = [
    "HP",
    "Dell",
    "Apple",
    "HP",
    "Dell"
]
```

## fit
- Machine ko data dikhana.
- Machine see only unique values.

## transform
- Machine khud mapping bana dega.
- Ye mapping alphabetically hoti hai.
-  for ex:-

|     Original     |             Result                 |
| ---------------- | ---------------------------------- |
| HP → ?           | Apple → 0                          |
| Dell → ?         | Dell → 1                           |
| Apple → ?        | HP → 2                             |

```python
encoded = le.fit_transform(company)
print(encoded)

''' company = [ "HP", "Dell", "Apple", "HP", "Dell" ]  original data
A = 0 , D = 1 , H = 2
''' 
```

```python
import pandas as pd
city = [ "Mumbai" ,  "Delhi" , "Uttar Pardesh" ]
le = LabelEncoder()
encoder = le.fit_transform(city)
print(encoder)
```

## Short
- fit() data se learn karta hai (unique categories identify karta hai).
  - Unique data values identify karna.
- transform() learned mapping ke according data ko convert karta hai.
- fit_transform() dono steps ek saath karta hai.


result = A = 0 , B = 1 etc
- Ye numbers sirf labels hain.
Inka mathematical meaning nahi hai.

```python
# Practice
from sklearn.preprocessing import LabelEncoder
skills = [ "Database" , "SQL" , "MySql" , "Java" , "Python" ]

le = LabelEncoder() # used for categorical data # creating object
encoded = le.fit_transform(skills)

print(encoded)
```

# One - Hot Encoding

## Method 1: pd.get_dummies()

```python
import pandas as pd

data = {
    "Company" : [ "HP" , "Dell" , "Apple" , "HP" ]
}

df = pd.DataFrame(data)

encoded_df = pd.get_dummies(df , dtype = int , drop_first = True)

# encoded_df = pd.get_dummies(df , dtype = int)

print(encoded_df)
```

```python
# drop_first = True ===> se we remove first column jiise machine per kam effect padega.
```

## Method 2: OneHotEncoder

```python
from sklearn.preprocessing import OneHotEncoder

company = [
    ["HP"],
    ["Dell"],
    ["Apple"],
    ["HP"],
    ["Apple"],
    ["Lenovo"]
]

encoder = OneHotEncoder()

encoded = encoder.fit_transform(company)

print(encoded.toarray())
```
