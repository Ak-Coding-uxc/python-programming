# Lesson 2 :- 
- Box Plot (sns.boxplot())
- Heatmap (sns.heatmap())
- Pair Plot (sns.pairplot())

   - they are use to understand ML project ka EDA (Exploratory Data Analysis)

```python
# 1. boxplot => used to detect outliers, minimum value, maximum value, median, outliers etc. 
import seaborn as sns
import matplotlib.pyplot as plt

# price = [ 30000, 40000, 5000000 , 60000 , 70000 , 80000]

price = [
    30000,
    35000,
    40000,
    45000,
    50000,
    55000,
    60000,
    65000,
    70000,
    900000
]

sns.boxplot(x = price)

plt.show()
```

```python
# 3. Pair Plot (read from chat gpt)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
'ram' : [4 , 8 , 16],
'ssd' : [256 , 512 , 512],
'price' : [30000 , 50000 , 80000]
}

df = pd.DataFrame(data)

# sns.pairplot(df)
sns.heatmap(df.corr()) # this crated a visual to understand.
plt.show()
```
