```python
import matplotlib.pyplot as plt

# Data
x = [12, 13, 14, 15, 16]
y = [25, 87, 63, 28, 20]

# Initialize a single figure for both subplots
plt.figure(figsize=(6, 8)) 
plt.suptitle("MY NEW GRAPH", fontsize=16, fontweight="bold")

# Subplot 1: Line and Scatter Plot
plt.subplot(2, 1, 1)
plt.plot(x, y, color="forestgreen", linewidth=2.5, marker='o', markersize=6, label="Marks")
plt.ylabel("Marks", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(loc="upper right")

# Subplot 2: Bar Chart
plt.subplot(2, 1, 2)
plt.bar(x, y, color="skyblue", edgecolor="navy", alpha=0.8)
plt.xlabel("Roll No", fontsize=11)
plt.ylabel("Marks", fontsize=11)
plt.grid(True, axis='y', linestyle="--", alpha=0.6)

# Optimize spacing and display the final figure once
plt.tight_layout()
plt.show()
```
