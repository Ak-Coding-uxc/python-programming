# Lesson 2 :- x and y labels , some plt.plot attributes etc

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[20,40,67,28,25]

plt.figure(figsize=(12,4)) # width , height
 # figure size upar aata h plot ke upar

plt.plot(x,y , color = "red" , linewidth = 5 , linestyle ="--") # its make graph

plt.title("Student Performance")

plt.xlabel("Student Roll Numbers")
plt.ylabel("Marks Obtained ( Out of 90 )")

plt.show() # its show graph
```

# Lesson 3 :- Types of Graphs

```python
import matplotlib.pyplot as plt

ram = [4 , 6 , 8 , 12]
price = [ 30000 , 50000 , 80000 , 140000]
plt.figure(figsize = (20,5))

plt.xlabel("Ram")
plt.ylabel("Price")
plt.title("Ram and their Price")  

# plt.plot(ram ,price)
#scatter graph => used to see trends
plt.scatter(ram,price)
plt.show()

# bar graph => used for comparison
plt.bar(ram,price)


plt.xlabel("Ram")
plt.ylabel("Price")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [12,14,16,18,21,22,24,25,27,30,31,35,40,42,45]

plt.hist(marks)

plt.show()
```

# Lesson 4:- Last Some Points

```python
import matplotlib.pyplot as plt

days = [ 1 , 2 , 3 ,4 , 5]

math = [10 , 20 , 30 , 40 , 56]
science = [ 24 , 78 , 45, 62 , 38]

plt.plot(days , math , label = "Math" , linestyle = ":" , color = "Red")
plt.plot(days , science , label = "Science" , linestyle = "--" , color = "Pink")
plt.legend() # to see label box

plt.show()
```

```python
# important
# - Line ke saath changes are attributes like color , linewidth , lable of line , linestyle etc
# - Graph ke saath changes are functions/methods of graph. like plt.xlabel() , plt.title() , plt.legend() , plt.figure(figsize = (8 , 9)) etc.


import matplotlib.pyplot as plt
# pyplot is the graph module of matplotlib

x = [1 , 1.5 , 3 , 4 ,5]
y = [ 10 , 20 , 30 , 40 , 50]
z = [ 20 , 30 , 40 , 60 , 100]
# this part makes graph
# This make only line
# plt.plot(x, y) 

#------
# Have other 3 types of graphs , 
# 1. Scatter Graph => Used to show trends , show only points not line (take 2 data ).
# 2. Bar Graph => Used to show comparison , take 2 data.
# 3. Histogram => Used to show distribution , take only 1 data. (ex:- how much in range 10 - 15)
# -- scatter graph , plt.plot can be combine and work together
#------

# plt.scatter(x,y)
# plt.bar(x,y , color = "GReen")
# plt.hist(x)


#  other functions of graphs

plt.figure(figsize = (10 , 5)) # width , height # size of the graph , write always before plotting 
plt.title("Z and Y") # Used to give title of graphs , can be write anywhere just before show
plt.grid() # Used to show gridlines


# some attributes of graphs ;- like color , linewidth , linestyle etc
plt.plot(x,z , label = "z")
plt.scatter(x,z) # only show dot.

plt.plot(x,y ,color = "pink" ,linewidth = 6 , linestyle = "--" , label = "y" )
# plt.scatter(x,y , color ="green" , linewidth = 10)
plt.legend() # used to show label , write only one time 

plt.xlabel("Kilogram")
plt.ylabel("Amount")


# this part show graph
plt.show()


```

```python
# Subplot => Ek hi figure mein 2 graphs

# SYNTAX => 
# plt.subplot(rows, columns, position) , position = 1 : first graph , 2: second graph etc

import matplotlib.pyplot as plt

x = [ 10 , 20 , 30 , 40 , 60]
y = [ 100 , 200 , 300 , 400 , 500]

plt.subplot(2 , 1 , 1) 

plt.plot(x,y)
plt.title(" Line Plot")


plt.subplot(2 , 1, 2)
plt.bar(x,y)
plt.title("Bar Graph")




# ---- जादूई लाइन जो ओवरलैपिंग ठीक करेगी ---- show se pahle likhna h.
plt.tight_layout() 

# used to save graph as images , write before show because after showing method graph close automatically
plt.savefig("graph.png")


plt.show()


```
